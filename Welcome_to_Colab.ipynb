# Part (2): Computing Self-Attention Scores

This notebook demonstrates how to compute self-attention scores for the sentence: **"What are the symptoms of diabetes?"** using the Scaled Dot-Product Attention formula:

$$\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V$$

```
import numpy as np
import pandas as pd

def softmax(x):
    e_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return e_x / e_x.sum(axis=-1, keepdims=True)

# 1. Define the input tokens
tokens = ["What", "are", "the", "symptoms", "of", "diabetes"]
seq_len = len(tokens)
d_model = 16  # Embedding dimension
d_k = 8       # Dimension of keys/queries

# 2. Represent tokens as random embeddings (for demonstration)
np.random.seed(42)
X = np.random.randn(seq_len, d_model)

# 3. Define weight matrices for Q, K, and V
W_Q = np.random.randn(d_model, d_k)
W_K = np.random.randn(d_model, d_k)
W_V = np.random.randn(d_model, d_k)

# 4. Compute Query, Key, and Value matrices
Q = X @ W_Q
K = X @ W_K
V = X @ W_V

# 5. Compute Scaled Dot-Product Attention Scores
# Score = Q * K^T / sqrt(d_k)
scores = (Q @ K.T) / np.sqrt(d_k)

# 6. Apply Softmax to get Attention Weights
attention_weights = softmax(scores)

# 7. Compute final output
output = attention_weights @ V

# Display the Attention Weights as a Heatmap DataFrame
weights_df = pd.DataFrame(attention_weights, index=tokens, columns=tokens)
print("Self-Attention Weights Matrix:")
display(weights_df.style.background_gradient(cmap='viridis'))
```
